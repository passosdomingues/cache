```python
# -*- coding: utf-8 -*-
"""
Análise Abrangente de Ameaças de Cibersegurança

Este script implementa um pipeline de Machine Learning de ponta a ponta,
seguindo rigorosamente a lista de verificação de 8 etapas para projetos de ML.
O objetivo é classificar o tipo de ameaça de cibersegurança do dataset
"Global Cybersecurity Threats (2015-2024)".

A estrutura é intencionalmente granular, com uma função para cada item
da lista de verificação, e narrada em primeira pessoa para simular o
processo de pensamento de uma IA enquanto executa a análise.
"""

# ==============================================================================
# 0. CONFIGURAÇÃO E IMPORTAÇÕES
# ==============================================================================
# Antes de tudo, preciso preparar meu ambiente. Vou instalar e importar todas as
# bibliotecas que precisarei para manipulação de dados, visualização e modelagem.
# É como arrumar minha bancada de trabalho antes de começar o projeto.
# ==============================================================================

# Instalar dependências necessárias silenciosamente
!pip install kagglehub -q
!pip install plotly -q
!pip install pandas numpy matplotlib seaborn scikit-learn xgboost lightgbm catboost -q

# Importar bibliotecas
import kagglehub
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
import os
import zipfile
import warnings
import joblib
import time
from IPython.display import display, HTML

# Ferramentas de Machine Learning do Scikit-learn
from sklearn.model_selection import train_test_split, cross_val_score, GridSearchCV, RandomizedSearchCV, StratifiedKFold
from sklearn.preprocessing import StandardScaler, OneHotEncoder, LabelEncoder, PowerTransformer
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline, FeatureUnion
from sklearn.impute import SimpleImputer
from sklearn.feature_selection import SelectKBest, f_classif
from sklearn.ensemble import RandomForestClassifier, GradientBoostingClassifier, VotingClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.naive_bayes import GaussianNB
from sklearn.metrics import classification_report, confusion_matrix, f1_score, accuracy_score, precision_score, recall_score
from sklearn.inspection import permutation_importance
import xgboost as xgb
import lightgbm as lgb
import catboost as cb

# Configurações para uma saída limpa e consistente
plt.style.use('seaborn-v0_8-whitegrid')
sns.set_palette("plasma")
warnings.filterwarnings('ignore')
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)

# Definir uma semente aleatória para garantir que os resultados sejam reproduzíveis
RANDOM_STATE = 42
np.random.seed(RANDOM_STATE)

# ==============================================================================
# DEFINIÇÃO DA CLASSE DO PIPELINE
# ==============================================================================
# Vou encapsular todo o processo em uma classe. Isso organiza o código,
# mantém o estado (como os dados e modelos) e torna o pipeline reutilizável.
# ==============================================================================

class CybersecurityThreatAnalysisPipeline:
    def __init__(self, kaggle_dataset_path="atharvasoundankar/global-cybersecurity-threats-2015-2024"):
        """
        Aqui eu inicializo todas as variáveis que vou usar ao longo da análise.
        É o meu setup inicial.
        """
        self.kaggle_dataset_path = kaggle_dataset_path
        self.data_frame = None
        self.X_train, self.X_test, self.y_train, self.y_test = None, None, None, None
        self.label_encoder_mapping = None
        self.preprocessing_pipeline = None
        self.model_comparison_results = None
        self.best_model_name = None
        self.tuned_model = None
        self.test_set_predictions = None
        self.final_f1_score = None
        self.feature_importances = None
        self.df_explore = None
        print("Tudo pronto. Iniciando o pipeline de análise de cibersegurança.")

    def run_full_pipeline(self):
        """
        Este é o meu método principal. Ele vai orquestrar a execução de cada
        passo, na ordem exata definida pela lista de verificação. Vamos começar.
        """
        # --- ETAPA 1: DEFINIR O PROBLEMA E VER O QUADRO GERAL ---
        self._step1_1_define_objective()
        self._step1_2_how_solution_will_be_used()
        self._step1_3_what_are_current_solutions()
        self._step1_4_how_to_frame_problem()
        self._step1_5_how_to_measure_performance()
        self._step1_6_is_performance_measure_aligned()
        self._step1_7_minimum_performance_needed()
        self._step1_8_what_are_comparable_problems()
        self._step1_9_is_human_expertise_available()
        self._step1_10_how_to_solve_manually()
        self._step1_11_list_assumptions()
        self._step1_12_verify_assumptions()

        # --- ETAPA 2: OBTER OS DADOS ---
        self._step2_1_list_data_needed()
        self._step2_2_find_and_document_data_source()
        self._step2_3_check_storage_space()
        self._step2_4_check_legal_obligations()
        self._step2_5_get_access_authorizations()
        self._step2_6_create_workspace()
        self._step2_7_get_the_data()
        self._step2_8_convert_data_format()
        self._step2_9_ensure_sensitive_info_is_protected()
        self._step2_10_check_size_and_type()
        self._step2_11_sample_a_test_set()

        # --- ETAPA 3: EXPLORAR OS DADOS ---
        self._step3_1_create_copy_for_exploration()
        self._step3_2_create_a_notebook()
        self._step3_3_study_each_attribute()
        self._step3_4_identify_target_attribute()
        self._step3_5_visualize_the_data()
        self._step3_6_study_correlations()
        self._step3_7_study_how_to_solve_manually()
        self._step3_8_identify_promising_transformations()
        self._step3_9_identify_extra_data()
        self._step3_10_document_what_i_learned_from_exploration()

        # --- ETAPA 4: PREPARAR OS DADOS ---
        self._step4_1_clean_the_data()
        self._step4_2_perform_feature_selection()
        self._step4_3_perform_feature_engineering()
        self._step4_4_perform_feature_scaling()

        # --- ETAPA 5: SELECIONAR MODELOS PROMISsores ---
        self._step5_1_train_quick_and_dirty_models()
        self._step5_2_measure_and_compare_performance()
        self._step5_3_analyze_most_significant_variables()
        self._step5_4_analyze_types_of_errors()
        self._step5_5_perform_quick_feature_selection_round()
        self._step5_6_perform_one_more_quick_iteration()
        self._step5_7_shortlist_top_models()

        # --- ETAPA 6: AJUSTAR FINAMENTE O SISTEMA ---
        self._step6_1_fine_tune_hyperparameters()
        self._step6_2_try_ensemble_methods()
        self._step6_3_measure_performance_on_test_set()

        # --- ETAPA 7: APRESENTAR A SOLUÇÃO ---
        self._step7_1_document_what_i_have_done()
        self._step7_2_create_a_nice_presentation()
        self._step7_3_explain_why_solution_achieves_objective()
        self._step7_4_present_interesting_points()
        self._step7_5_ensure_key_findings_are_communicated()

        # --- ETAPA 8: LANÇAR, MONITORAR E MANTER ---
        self._step8_1_get_solution_ready_for_production()
        self._step8_2_write_monitoring_code()
        self._step8_3_retrain_models_on_a_regular_basis()

        print("\n\n" + "="*80 + "\nPIPELINE CONCLUÍDO COM SUCESSO!\n" + "="*80)
        print(f"Desempenho final no conjunto de teste: F1-Score = {self.final_f1_score:.4f}")

    # ==============================================================================
    # ETAPA 1: DEFINIR O PROBLEMA E VER O QUADRO GERAL
    # ==============================================================================
    def _step1_1_define_objective(self):
        print("\n--- ETAPA 1.1: Definindo o objetivo em termos de negócios ---")
        print("Estou pensando sobre o que o negócio realmente precisa. O objetivo principal é criar um sistema que classifique automaticamente o tipo de ciberataque ('Attack Type') com base em suas características. Isso ajudará as equipes de segurança a responderem mais rapidamente e de forma mais eficaz, priorizando os incidentes mais críticos e direcionando-os para os especialistas corretos.")

    def _step1_2_how_solution_will_be_used(self):
        print("\n--- ETAPA 1.2: Como a solução será usada? ---")
        print("Estou imaginando o cenário de uso. Esta solução será integrada ao sistema de gerenciamento de segurança (SIEM) da organização. Quando um novo alerta de incidente for gerado, meu modelo automaticamente categorizará o tipo de ataque, ajudando os analistas a priorizar sua resposta com base na criticidade e no tipo de ameaça.")

    def _step1_3_what_are_current_solutions(self):
        print("\n--- ETAPA 1.3: Quais são as soluções atuais? ---")
        print("Analisando a situação atual: esse processo é predominantemente manual. Um analista de segurança precisa investigar cada alerta para classificá-lo, o que consome tempo valioso e está sujeito a erros humanos. Minha solução visa automatizar essa triagem inicial, liberando os analistas para tarefas mais complexas.")

    def _step1_4_how_to_frame_problem(self):
        print("\n--- ETAPA 1.4: Como devo enquadrar este problema? ---")
        print("Estou analisando a natureza do problema. Tenho dados históricos com rótulos (os tipos de ataque já estão definidos). O alvo tem várias categorias (Malware, Phishing, DDoS, etc.). Portanto, este é claramente um problema de **classificação multiclasse supervisionada**. O treinamento será 'offline' usando dados históricos, mas poderia evoluir para um sistema online com atualizações contínuas.")

    def _step1_5_how_to_measure_performance(self):
        print("\n--- ETAPA 1.5: Como o desempenho deve ser medido? ---")
        print("Pensando nas métricas adequadas: a acurácia é uma métrica inicial útil, mas os tipos de ataque podem ser desbalanceados (alguns mais comuns que outros). Por isso, o **F1-Score Ponderado (Weighted F1-Score)** é mais apropriado. Ele mede o equilíbrio entre precisão e recall para cada classe e pondera pela sua frequência, dando uma visão mais justa do desempenho geral. Também considerarei a matriz de confusão para entender os erros específicos.")

    def _step1_6_is_performance_measure_aligned(self):
        print("\n--- ETAPA 1.6: A medida de desempenho está alinhada com o objetivo? ---")
        print("Verificando o alinhamento: Sim, completamente. Um F1-Score alto significa que o modelo é bom em identificar corretamente a maioria dos ataques sem gerar muitos alarmes falsos. Isso está diretamente alinhado com o objetivo de negócio de fornecer uma ferramenta de triagem confiável que ajude os analistas a priorizar eficientemente.")

    def _step1_7_minimum_performance_needed(self):
        print("\n--- ETAPA 1.7: Qual o desempenho mínimo necessário? ---")
        print("Estabelecendo um benchmark: Para ser realmente útil e superar um processo manual, eu miraria em um F1-Score Ponderado acima de 85%. Abaixo disso, a confiança na ferramenta pode ser baixa e os analistas podem preferir confiar em sua própria avaliação.")

    def _step1_8_what_are_comparable_problems(self):
        print("\n--- ETAPA 1.8: Quais são problemas comparáveis? ---")
        print("Buscando analogias: Isso é semelhante à categorização de tickets de suporte, classificação de e-mails (spam, social, promoções) ou diagnóstico médico com base em sintomas. Posso reutilizar técnicas e modelos comuns a esses problemas, como Random Forests, Gradient Boosting e abordagens de ensemble.")

    def _step1_9_is_human_expertise_available(self):
        print("\n--- ETAPA 1.9: Há expertise humana disponível? ---")
        print("Avaliando recursos especializados: Sim, os próprios dados foram rotulados por especialistas. Em um projeto real, eu consultaria analistas de cibersegurança para validar minhas descobertas e entender melhor as nuances dos dados e do domínio.")

    def _step1_10_how_to_solve_manually(self):
        print("\n--- ETAPA 1.10: Como eu resolveria o problema manualmente? ---")
        print("Simulando o processo manual: Um analista examinaria a indústria-alvo, a fonte do ataque, o número de usuários afetados, o ano e outros fatores para identificar padrões. Por exemplo, um ataque a muitos usuários no setor financeiro com uma fonte de email pode ser classificado como 'Phishing'. Meu modelo aprenderá esses padrões em escala.")

    def _step1_11_list_assumptions(self):
        print("\n--- ETAPA 1.11: Listando as suposições ---")
        print("Identificando premissas: Estou assumindo que: 1) As features fornecidas são suficientes para prever o tipo de ataque. 2) Os rótulos no dataset são precisos e consistentes. 3) O padrão de ataques passados (2015-2024) ainda é relevante para ataques futuros. 4) Não há vazamento de dados entre features e target.")

    def _step1_12_verify_assumptions(self):
        print("\n--- ETAPA 1.12: Verificando as suposições ---")
        print("Validando premissas: Não posso verificar completamente a precisão dos rótulos, então confiarei na fonte do dataset. A relevância das features será avaliada durante a exploração e modelagem. A suposição sobre a relevância dos padrões passados é uma limitação que precisará ser monitorada após a implantação.")

    # ==============================================================================
    # ETAPA 2: OBTER OS DADOS
    # ==============================================================================
    def _step2_1_list_data_needed(self):
        print("\n--- ETAPA 2.1: Listando os dados de que preciso ---")
        print("Definindo requisitos de dados: Preciso do dataset 'Global Cybersecurity Threats (2015-2024)'. Ele parece ter cerca de 3000 amostras com features como tipo de ataque, indústria-alvo, país, fonte do ataque, etc. Este tamanho é adequado para um protótipo inicial.")

    def _step2_2_find_and_document_data_source(self):
        print("\n--- ETAPA 2.2: Encontrando e documentando a fonte dos dados ---")
        print(f"Localizando a fonte: A fonte é o Kaggle, no caminho: {self.kaggle_dataset_path}. Vou usar a API do KaggleHub para baixar os dados de forma automatizada e reprodutível.")

    def _step2_3_check_storage_space(self):
        print("\n--- ETAPA 2.3: Verificando o espaço de armazenamento ---")
        print("Avaliando requisitos de armazenamento: O dataset é pequeno, na ordem de kilobytes. O espaço não será um problema neste ambiente, mas é sempre bom verificar para datasets maiores.")

    def _step2_4_check_legal_obligations(self):
        print("\n--- ETAPA 2.4: Verificando obrigações legais ---")
        print("Considerando aspectos legais: O dataset parece ser público e destinado a fins de pesquisa e educação. Nenhuma autorização especial parece ser necessária, mas sempre é importante verificar os termos de uso.")

    def _step2_5_get_access_authorizations(self):
        print("\n--- ETAPA 2.5: Obtendo autorizações de acesso ---")
        print("Obtendo acesso: Meu ambiente está configurado com as credenciais necessárias para acessar a API do Kaggle. Tenho autorização para baixar os dados.")

    def _step2_6_create_workspace(self):
        print("\n--- ETAPA 2.6: Criando um espaço de trabalho ---")
        print("Preparando o ambiente: Este notebook Colab é meu espaço de trabalho principal. Ele tem armazenamento suficiente e poder de computação adequado para esta análise. Vou organizar pastas para dados, modelos e resultados.")

    def _step2_7_get_the_data(self):
        print("\n--- ETAPA 2.7: Obtendo os dados ---")
        print("Iniciando o download dos dados: Agora vou baixar os dados. O processo precisa ser robusto, pois o Kaggle pode fornecer um arquivo zip ou um diretório.")
        try:
            path = kagglehub.dataset_download(self.kaggle_dataset_path)
            csv_path = None

            if os.path.isdir(path):
                print("   Recebi um diretório. Procurando o arquivo CSV dentro dele...")
                csv_files = [f for f in os.listdir(path) if f.endswith('.csv')]
                if not csv_files:
                    raise FileNotFoundError("Nenhum CSV encontrado no diretório.")
                csv_path = os.path.join(path, csv_files[0])
            elif os.path.isfile(path) and path.endswith('.zip'):
                print("   Recebi um arquivo zip. Descompactando...")
                with zipfile.ZipFile(path, 'r') as z:
                    csv_files = [name for name in z.namelist() if name.endswith('.csv')]
                    if not csv_files:
                        raise FileNotFoundError("Nenhum CSV encontrado no arquivo zip.")
                    csv_filename = csv_files[0]
                    z.extractall("dataset_temp")
                    csv_path = os.path.join("dataset_temp", csv_filename)
            else:
                raise ValueError("Formato de dados não reconhecido.")

            print(f"   Carregando dados de '{csv_path}'...")
            self.data_frame = pd.read_csv(csv_path)
            print(f"   Dados carregados com sucesso! Shape: {self.data_frame.shape}")

        except Exception as e:
            print(f"   Falha ao obter os dados: {e}")
            # Fallback: tentar carregar de um caminho local se disponível
            try:
                print("   Tentando carregar de arquivo local...")
                self.data_frame = pd.read_csv("global_cybersecurity_threats.csv")
                print("   Dados carregados do arquivo local com sucesso!")
            except:
                raise RuntimeError("Não foi possível carregar os dados de nenhuma fonte.")

    def _step2_8_convert_data_format(self):
        print("\n--- ETAPA 2.8: Convertendo os dados para um formato manipulável ---")
        print("Preparando os dados para análise: Os dados foram carregados como um DataFrame do Pandas, que é o formato ideal para manipulação e análise. Não preciso de conversões adicionais neste momento.")

    def _step2_9_ensure_sensitive_info_is_protected(self):
        print("\n--- ETAPA 2.9: Garantindo a proteção de informações sensíveis ---")
        print("Verificando dados sensíveis: Examinando as colunas, vejo que temos 'Country', 'Target Industry', mas não há informações pessoais identificáveis (PII) como nomes, emails ou números de documento. O dataset parece seguro e anonimizado por natureza.")

    def _step2_10_check_size_and_type(self):
        print("\n--- ETAPA 2.10: Verificando o tamanho e o tipo dos dados ---")
        print(f"Analisando a estrutura dos dados: Os dados são tabulares, com {self.data_frame.shape[0]} linhas (amostras) e {self.data_frame.shape[1]} colunas (features). É um conjunto de amostras independentes, não uma série temporal.")

    def _step2_11_sample_a_test_set(self):
        print("\n--- ETAPA 2.11: Separando um conjunto de teste ---")
        print("Prevenindo data snooping: Este é um passo crucial. Vou separar 20% dos dados para um conjunto de teste agora mesmo. Vou guardá-lo e **nunca** olhar para ele até a avaliação final. Isso evita vazamento de informação e me dá uma estimativa real do desempenho do modelo.")

        target_col = 'Attack Type'
        if target_col not in self.data_frame.columns:
            # Tentar encontrar a coluna target por padrão
            possible_targets = [col for col in self.data_frame.columns if 'attack' in col.lower() or 'type' in col.lower()]
            if possible_targets:
                target_col = possible_targets[0]
                print(f"   Coluna alvo não encontrada. Usando '{target_col}' como alvo.")
            else:
                raise ValueError("Não foi possível identificar a coluna alvo.")

        X = self.data_frame.drop(columns=[target_col])
        y = self.data_frame[target_col]

        # Usar stratify para garantir que a proporção de cada tipo de ataque
        # seja a mesma nos conjuntos de treino e teste
        self.X_train, self.X_test, self.y_train, self.y_test = train_test_split(
            X, y, test_size=0.2, random_state=RANDOM_STATE, stratify=y
        )
        print(f"   Dados divididos. Treino: {len(self.y_train)} amostras. Teste: {len(self.y_test)} amostras. O conjunto de teste está agora 'trancado'.")

    # ==============================================================================
    # ETAPA 3: EXPLORAR OS DADOS
    # ==============================================================================
    def _step3_1_create_copy_for_exploration(self):
        print("\n--- ETAPA 3.1: Criando uma cópia dos dados para exploração ---")
        print("Preservando dados originais: Vou trabalhar apenas com uma cópia dos dados de treino para não afetar os originais. Segurança em primeiro lugar.")
        self.df_explore = self.X_train.copy()
        self.df_explore['Attack Type'] = self.y_train.copy()

    def _step3_2_create_a_notebook(self):
        print("\n--- ETAPA 3.2: Criando um notebook para registrar a exploração ---")
        print("Documentando a análise: Este script funciona como meu notebook de exploração. Cada passo e cada descoberta serão documentados aqui mesmo para referência futura.")

    def _step3_3_study_each_attribute(self):
        print("\n--- ETAPA 3.3: Estudando cada atributo e suas características ---")
        print("Realizando análise exploratória inicial: Vou examinar os tipos de dados, valores ausentes e estatísticas descritivas para entender melhor cada feature.")

        print("\n   Informações gerais sobre o dataset:")
        self.df_explore.info()

        print("\n   Estatísticas descritivas para variáveis numéricas:")
        display(self.df_explore.describe())

        print("\n   Estatísticas descritivas para variáveis categóricas:")
        categorical_cols = self.df_explore.select_dtypes(include=['object']).columns
        for col in categorical_cols:
            print(f"\n   {col}:")
            print(f"   Número de categorias: {self.df_explore[col].nunique()}")
            print(f"   Categorias mais frequentes:")
            display(self.df_explore[col].value_counts().head())

        print("\n   Valores ausentes por coluna:")
        missing_data = self.df_explore.isnull().sum()
        missing_percent = (missing_data / len(self.df_explore)) * 100
        missing_table = pd.DataFrame({'Valores Ausentes': missing_data, 'Percentual': missing_percent})
        display(missing_table[missing_table['Valores Ausentes'] > 0])

        print("\n   Insights iniciais:")
        print("   - Temos colunas numéricas ('Year', 'Number of Affected Users') e categóricas.")
        print("   - Algumas colunas têm valores ausentes que precisarão ser tratados.")
        print("   - A coluna 'Country' tem alta cardinalidade (muitas categorias únicas).")

    def _step3_4_identify_target_attribute(self):
        print("\n--- ETAPA 3.4: Identificando o(s) atributo(s) alvo ---")
        print("Definindo o objetivo da modelagem: Como definido na Etapa 1, meu alvo é a coluna **'Attack Type'**. É ela que eu quero prever. Vou examinar sua distribuição.")

        target_counts = self.df_explore['Attack Type'].value_counts()
        target_percent = (target_counts / len(self.df_explore)) * 100

        print("\n   Distribuição do target 'Attack Type':")
        for attack_type, count, percent in zip(target_counts.index, target_counts.values, target_percent.values):
            print(f"   {attack_type}: {count} amostras ({percent:.2f}%)")

        print(f"\n   Total de classes: {len(target_counts)}")
        print("   O dataset está desbalanceado, o que é comum em problemas do mundo real de segurança.")

    def _step3_5_visualize_the_data(self):
        print("\n--- ETAPA 3.5: Visualizando os dados ---")
        print("Criando visualizações para melhor compreensão dos dados:")

        # 1. Distribuição do target
        print("\n   1. Distribuição dos Tipos de Ataque:")
        plt.figure(figsize=(10, 6))
        ax = sns.countplot(data=self.df_explore, y='Attack Type', order=self.df_explore['Attack Type'].value_counts().index)
        plt.title('Distribuição dos Tipos de Ataque')
        plt.xlabel('Contagem')
        for p in ax.patches:
            ax.annotate(f'{int(p.get_width())}', (p.get_width(), p.get_y() + p.get_height()/2),
                       ha='left', va='center')
        plt.tight_layout()
        plt.show()

        # 2. Distribuição por ano
        print("\n   2. Distribuição de Ataques por Ano:")
        if 'Year' in self.df_explore.columns:
            plt.figure(figsize=(10, 6))
            year_counts = self.df_explore['Year'].value_counts().sort_index()
            sns.barplot(x=year_counts.index, y=year_counts.values)
            plt.title('Número de Ataques por Ano')
            plt.xlabel('Ano')
            plt.ylabel('Contagem')
            plt.xticks(rotation=45)
            plt.tight_layout()
            plt.show()

        # 3. Distribuição por indústria
        print("\n   3. Top 10 Indústrias Mais Atacadas:")
        if 'Target Industry' in self.df_explore.columns:
            plt.figure(figsize=(10, 6))
            industry_counts = self.df_explore['Target Industry'].value_counts().nlargest(10)
            sns.barplot(x=industry_counts.values, y=industry_counts.index)
            plt.title('Top 10 Indústrias Mais Atacadas')
            plt.xlabel('Contagem')
            plt.tight_layout()
            plt.show()

        # 4. Distribuição de usuários afetados
        print("\n   4. Distribuição do Número de Usuários Afetados:")
        if 'Number of Affected Users' in self.df_explore.columns:
            plt.figure(figsize=(10, 6))
            # Usar escala logarítmica para melhor visualização se houver outliers
            if self.df_explore['Number of Affected Users'].max() > 10 * self.df_explore['Number of Affected Users'].median():
                plt.hist(np.log1p(self.df_explore['Number of Affected Users']), bins=30)
                plt.xlabel('Log(Number of Affected Users + 1)')
            else:
                plt.hist(self.df_explore['Number of Affected Users'], bins=30)
                plt.xlabel('Number of Affected Users')
            plt.ylabel('Frequência')
            plt.title('Distribuição do Número de Usuários Afetados')
            plt.tight_layout()
            plt.show()

    def _step3_6_study_correlations(self):
        print("\n--- ETAPA 3.6: Estudando as correlações entre atributos ---")
        print("Analisando relações entre variáveis:")

        # Para variáveis numéricas
        numeric_cols = self.df_explore.select_dtypes(include=np.number).columns
        if len(numeric_cols) > 1:
            print("\n   Matriz de correlação para variáveis numéricas:")
            corr_matrix = self.df_explore[numeric_cols].corr()
            plt.figure(figsize=(10, 8))
            sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', center=0, fmt='.2f')
            plt.title('Matriz de Correlação - Variáveis Numéricas')
            plt.tight_layout()
            plt.show()

        # Para variáveis categóricas (usando cramers V seria melhor, mas é mais complexo)
        # Aqui vou fazer uma análise simplificada
        print("\n   Relações entre variáveis categóricas e o target:")
        categorical_cols = self.df_explore.select_dtypes(include=['object']).columns
        categorical_cols = [col for col in categorical_cols if col != 'Attack Type']

        for col in categorical_cols[:3]:  # Analisar apenas as primeiras 3 para não sobrecarregar
            if self.df_explore[col].nunique() <= 15:  # Só plotar se não tiver muitas categorias
                print(f"\n   Distribuição de {col} por Tipo de Ataque:")
                plt.figure(figsize=(12, 6))

                # Calcular proporções
                cross_tab = pd.crosstab(self.df_explore[col], self.df_explore['Attack Type'])
                cross_tab_prop = cross_tab.div(cross_tab.sum(axis=1), axis=0)

                cross_tab_prop.plot(kind='bar', stacked=True)
                plt.title(f'Proporção de Tipos de Ataque por {col}')
                plt.ylabel('Proporção')
                plt.legend(title='Attack Type', bbox_to_anchor=(1.05, 1), loc='upper left')
                plt.tight_layout()
                plt.show()

    def _step3_7_study_how_to_solve_manually(self):
        print("\n--- ETAPA 3.7: Estudando como resolver o problema manualmente ---")
        print("Pensando como um analista humano: Revisitando essa ideia, um analista provavelmente criaria regras heurísticas baseadas em padrões conhecidos:")
        print("   - 'SE a indústria é Financeira E a fonte é Email, ENTÃO alta probabilidade de Phishing'")
        print("   - 'SE há muitos usuários afetados E a fonte é múltiplos IPs, ENTÃO possivelmente DDoS'")
        print("   - 'SE o país é conhecido por ataques avançados E a indústria é Governo, ENTÃO possivelmente APT'")
        print("Modelos como Árvores de Decisão e Random Forest automatizam exatamente esse processo, mas de forma mais sofisticada e em maior escala.")

    def _step3_8_identify_promising_transformations(self):
        print("\n--- ETAPA 3.8: Identificando transformações promissoras ---")
        print("Planejando o pré-processamento: Com base na minha exploração, a estratégia de preparação de dados está se tornando clara:")
        print("   1. **Tratamento de Valores Ausentes**: Preencher com a moda para categóricas e mediana para numéricas.")
        print("   2. **Colunas Numéricas**: 'Year' e 'Number of Affected Users' precisam ser escalonadas. Como 'Number of Affected Users' tem distribuição assimétrica, considerarei aplicar log transform.")
        print("   3. **Colunas Categóricas**: 'Target Industry', 'Attack Source', etc. precisarão de encoding. One-Hot Encoding pode ser apropriado para categorias com poucos valores, enquanto Target Encoding pode ser melhor para categorias com muitos valores.")
        print("   4. **Colunas com Alta Cardinalidade**: 'Country' tem muitas categorias únicas. Considerarei agrupar países por região ou usar técnicas de encoding mais avançadas.")

    def _step3_9_identify_extra_data(self):
        print("\n--- ETAPA 3.9: Identificando dados extras que seriam úteis ---")
        print("Considerando aprimoramentos futuros: Mais dados sempre melhoram os modelos! Dados mais recentes (pós-2024) seriam valiosos. Informações adicionais como:")
        print("   - Vetores de ataque específicos (tipo de malware, técnica de phishing)")
        print("   - Dados temporais mais granulares (mês, dia da semana, hora)")
        print("   - Informações geográficas mais detalhadas (região, densidade populacional)")
        print("   - Dados sobre contramedidas implementadas")
        print("Tornariam o modelo ainda mais preciso e robusto.")

    def _step3_10_document_what_i_learned_from_exploration(self):
        print("\n--- ETAPA 3.10: Documentando o que aprendi ---")
        print("Resumo da exploração e insights principais:")
        print("   1. O problema é uma classificação multiclasse com classes desbalanceadas.")
        print("   2. Os dados contêm features numéricas e categóricas com alguns valores ausentes.")
        print("   3. A coluna 'Country' tem alta cardinalidade e precisará de tratamento especial.")
        print("   4. Existem relações discerníveis entre algumas features e o target.")
        print("   5. A estratégia de pré-processamento envolverá:")
        print("      - Imputação de valores ausentes")
        print("      - Transformação de variáveis numéricas (possivelmente log transform)")
        print("      - Encoding apropriado de variáveis categóricas")
        print("      - Escalonamento de features")

    # ==============================================================================
    # ETAPA 4: PREPARAR OS DADOS
    # ==============================================================================
    def _step4_1_clean_the_data(self):
        print("\n--- ETAPA 4.1: Limpando os dados ---")
        print("Iniciando a limpeza dos dados: Vou tratar valores ausentes e decidir sobre a remoção de colunas problemáticas.")

        # Análise inicial de valores ausentes
        missing_before = self.X_train.isnull().sum()
        print("   Valores ausentes antes da limpeza:")
        for col, count in missing_before.items():
            if count > 0:
                percent = (count / len(self.X_train)) * 100
                print(f"   {col}: {count} valores ausentes ({percent:.2f}%)")

        # Decisões sobre colunas com muitos valores ausentes
        cols_to_drop = []
        for col in self.X_train.columns:
            missing_count = self.X_train[col].isnull().sum()
            missing_percent = (missing_count / len(self.X_train)) * 100

            if missing_percent > 30:  # Se mais de 30% dos valores estão ausentes
                print(f"   Decisão: Removendo coluna '{col}' com {missing_percent:.2f}% de valores ausentes")
                cols_to_drop.append(col)
            elif missing_percent > 0:
                print(f"   Decisão: Manter coluna '{col}' e imputar valores ausentes ({missing_percent:.2f}%)")

        # Remover colunas com muitos valores ausentes
        if cols_to_drop:
            self.X_train = self.X_train.drop(columns=cols_to_drop)
            self.X_test = self.X_test.drop(columns=cols_to_drop)
            print(f"   Colunas removidas: {cols_to_drop}")

        # Também considerar remover colunas com alta cardinalidade
        high_cardinality_cols = []
        for col in self.X_train.select_dtypes(include=['object']).columns:
            if self.X_train[col].nunique() > 50:  # Limite arbitrário para alta cardinalidade
                high_cardinality_cols.append(col)
                print(f"   Coluna '{col}' tem alta cardinalidade: {self.X_train[col].nunique()} categorias únicas")

        if high_cardinality_cols:
            # Para este exemplo, vou remover colunas com alta cardinalidade
            # Em um cenário real, consideraria técnicas como target encoding
            self.X_train = self.X_train.drop(columns=high_cardinality_cols)
            self.X_test = self.X_test.drop(columns=high_cardinality_cols)
            print(f"   Colunas com alta cardinalidade removidas: {high_cardinality_cols}")

        print("   Limpeza inicial concluída.")

    def _step4_2_perform_feature_selection(self):
        print("\n--- ETAPA 4.2: Realizando seleção de features ---")
        print("Avaliando a importância das features: A remoção das colunas na etapa anterior já foi uma forma de seleção manual. Agora vou usar uma abordagem mais estatística.")

        # Primeiro, preciso preparar os dados para análise de feature importance
        # Vou criar um pré-processamento básico para lidar com valores ausentes e encoding
        numeric_features = self.X_train.select_dtypes(include=np.number).columns.tolist()
        categorical_features = self.X_train.select_dtypes(exclude=np.number).columns.tolist()

        # Pré-processadores básicos
        numeric_transformer = Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())])

        categorical_transformer = Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))])

        preprocessor = ColumnTransformer(
            transformers=[
                ('num', numeric_transformer, numeric_features),
                ('cat', categorical_transformer, categorical_features)])

        # Aplicar pré-processamento
        X_train_processed = preprocessor.fit_transform(self.X_train)

        # Obter nomes das features após one-hot encoding
        feature_names = numeric_features.copy()
        ohe = preprocessor.named_transformers_['cat'].named_steps['onehot']
        categorical_features_encoded = ohe.get_feature_names_out(categorical_features)
        feature_names.extend(categorical_features_encoded)

        # Usar RandomForest para obter importância das features
        rf = RandomForestClassifier(n_estimators=100, random_state=RANDOM_STATE)
        rf.fit(X_train_processed, self.y_train)

        # Criar DataFrame com importâncias
        feature_importances = pd.DataFrame({
            'feature': feature_names,
            'importance': rf.feature_importances_
        }).sort_values('importance', ascending=False)

        # Selecionar as top-K features (por exemplo, top 20)
        top_k = 20
        top_features = feature_importances.head(top_k)['feature'].tolist()

        print(f"   Top {top_k} features mais importantes:")
        for i, (feature, importance) in enumerate(zip(feature_importances['feature'], feature_importances['importance'])):
            if i < top_k:
                print(f"   {i+1}. {feature}: {importance:.4f}")

        # Em um cenário real, eu usaria essas informações para selecionar features
        # Por simplicidade, vou manter todas as features por agora
        self.feature_importances = feature_importances
        print("   Feature selection concluído (informacional por enquanto).")

    def _step4_3_perform_feature_engineering(self):
        print("\n--- ETAPA 4.3: Realizando engenharia de features ---")
        print("Criando novas features: Vou adicionar algumas features derivadas que podem ser úteis para o modelo.")

        # Exemplo: criar uma feature binária para indicar se o número de usuários afetados está acima da média
        if 'Number of Affected Users' in self.X_train.columns:
            user_median = self.X_train['Number of Affected Users'].median()
            self.X_train['High_Impact'] = (self.X_train['Number of Affected Users'] > user_median).astype(int)
            self.X_test['High_Impact'] = (self.X_test['Number of Affected Users'] > user_median).astype(int)
            print("   Feature criada: High_Impact (ataques com número de usuários afetados acima da mediana)")

        # Exemplo: extrair década do ano se a coluna Year existir
        if 'Year' in self.X_train.columns:
            self.X_train['Decade'] = (self.X_train['Year'] // 10) * 10
            self.X_test['Decade'] = (self.X_test['Year'] // 10) * 10
            print("   Feature criada: Decade (década do ano)")

        print("   Engenharia de features básica concluída.")

    def _step4_4_perform_feature_scaling(self):
        print("\n--- ETAPA 4.4: Realizando escalonamento e transformação de features ---")
        print("Construindo o pipeline de pré-processamento completo: Agora vou criar o pipeline que será usado no treinamento do modelo.")

        # Identificar features numéricas e categóricas
        numeric_features = self.X_train.select_dtypes(include=np.number).columns.tolist()
        categorical_features = self.X_train.select_dtypes(exclude=np.number).columns.tolist()

        print(f"   Features numéricas: {numeric_features}")
        print(f"   Features categóricas: {categorical_features}")

        # Pré-processadores
        numeric_transformer = Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='median')),
            ('scaler', StandardScaler())])

        categorical_transformer = Pipeline(steps=[
            ('imputer', SimpleImputer(strategy='most_frequent')),
            ('onehot', OneHotEncoder(handle_unknown='ignore', sparse_output=False))])

        # Combinar pré-processadores
        self.preprocessing_pipeline = ColumnTransformer(
            transformers=[
                ('num', numeric_transformer, numeric_features),
                ('cat', categorical_transformer, categorical_features)])

        # Aplicar o pré-processamento
        print("   Ajustando o pré-processador com os dados de TREINO...")
        self.X_train_prepared = self.preprocessing_pipeline.fit_transform(self.X_train)
        print("   Aplicando a MESMA transformação nos dados de TESTE...")
        self.X_test_prepared = self.preprocessing_pipeline.transform(self.X_test)

        # Codificar as labels (transformar de texto para números)
        le = LabelEncoder()
        self.y_train_encoded = le.fit_transform(self.y_train)
        self.y_test_encoded = le.transform(self.y_test)
        self.label_encoder_mapping = dict(zip(le.classes_, le.transform(le.classes_)))

        print(f"   Pré-processamento concluído. Dimensão dos dados de treino: {self.X_train_prepared.shape}")
        print(f"   Mapeamento do target: {self.label_encoder_mapping}")

    # ==============================================================================
    # ETAPA 5: SELECIONAR MODELOS PROMISsores
    # ==============================================================================
    def _step5_1_train_quick_and_dirty_models(self):
        print("\n--- ETAPA 5.1: Treinando modelos rápidos de diferentes categorias ---")
        print("Iniciando a comparação de modelos: Vou treinar rapidamente vários modelos com configurações padrão para identificar os mais promissores.")

        # Definir os modelos a serem testados
        self.models_to_test = {
            'Logistic Regression': LogisticRegression(random_state=RANDOM_STATE, max_iter=1000),
            'Random Forest': RandomForestClassifier(random_state=RANDOM_STATE, n_jobs=-1),
            'Gradient Boosting': GradientBoostingClassifier(random_state=RANDOM_STATE),
            'SVM': SVC(random_state=RANDOM_STATE, probability=True),
            'Naive Bayes': GaussianNB(),
            'XGBoost': xgb.XGBClassifier(random_state=RANDOM_STATE, use_label_encoder=False, eval_metric='mlogloss'),
            'LightGBM': lgb.LGBMClassifier(random_state=RANDOM_STATE, n_jobs=-1),
            'CatBoost': cb.CatBoostClassifier(random_state=RANDOM_STATE, verbose=0)
        }

        print(f"   Modelos a serem testados: {list(self.models_to_test.keys())}")

    def _step5_2_measure_and_compare_performance(self):
        print("\n--- ETAPA 5.2: Medindo e comparando o desempenho ---")
        print("Avaliando modelos com validação cruzada: Vou usar validação cruzada estratificada para obter uma estimativa robusta do desempenho de cada modelo.")

        results = {}
        fit_times = {}
        kf = StratifiedKFold(n_splits=5, shuffle=True, random_state=RANDOM_STATE)

        for name, model in self.models_to_test.items():
            print(f"   Avaliando {name}...")
            start_time = time.time()

            try:
                # Usar cross_val_score para obter scores de validação cruzada
                cv_scores = cross_val_score(model, self.X_train_prepared, self.y_train_encoded,
                                          cv=kf, scoring='f1_weighted', n_jobs=-1)
                fit_time = time.time() - start_time

                results[name] = {
                    'mean_f1': np.mean(cv_scores),
                    'std_f1': np.std(cv_scores),
                    'fit_time': fit_time
                }

                print(f"   {name}: F1-Score = {np.mean(cv_scores):.4f} (±{np.std(cv_scores):.4f}), Tempo = {fit_time:.2f}s")

            except Exception as e:
                print(f"   Erro ao avaliar {name}: {e}")
                results[name] = {'mean_f1': 0, 'std_f1': 0, 'fit_time': 0}

        # Criar DataFrame com resultados
        self.model_comparison_results = pd.DataFrame.from_dict(results, orient='index')
        self.model_comparison_results = self.model_comparison_results.sort_values('mean_f1', ascending=False)

        print("\n   Resultados da comparação de modelos:")
        display(self.model_comparison_results)

        # Visualizar resultados
        plt.figure(figsize=(12, 6))
        models = self.model_comparison_results.index
        means = self.model_comparison_results['mean_f1']
        stds = self.model_comparison_results['std_f1']

        plt.bar(range(len(models)), means, yerr=stds, capsize=5)
        plt.xlabel('Modelos')
        plt.ylabel('F1-Score (Weighted)')
        plt.title('Comparação de Desempenho dos Modelos')
        plt.xticks(range(len(models)), models, rotation=45, ha='right')
        plt.tight_layout()
        plt.show()

    def _step5_3_analyze_most_significant_variables(self):
        print("\n--- ETAPA 5.3: Analisando as variáveis mais significativas ---")
        print("Examinando a importância das features: Vou treinar o melhor modelo e analisar quais features são mais importantes para suas previsões.")

        # Treinar o melhor modelo em todos os dados de treino
        best_model_name = self.model_comparison_results.index[0]
        best_model = self.models_to_test[best_model_name]

        print(f"   Treinando {best_model_name} para análise de importância de features...")
        best_model.fit(self.X_train_prepared, self.y_train_encoded)

        # Obter importância das features
        if hasattr(best_model, 'feature_importances_'):
            # Para modelos baseados em árvores
            importances = best_model.feature_importances_

            # Obter nomes das features após pré-processamento
            numeric_features = self.X_train.select_dtypes(include=np.number).columns.tolist()
            categorical_features = self.X_train.select_dtypes(exclude=np.number).columns.tolist()

            # Para features categóricas one-hot encoded, precisamos dos nomes expandidos
            ohe = self.preprocessing_pipeline.named_transformers_['cat'].named_steps['onehot']
            categorical_features_encoded = ohe.get_feature_names_out(categorical_features)

            all_feature_names = numeric_features + list(categorical_features_encoded)

            # Criar DataFrame com importâncias
            feature_importance_df = pd.DataFrame({
                'feature': all_feature_names,
                'importance': importances
            }).sort_values('importance', ascending=False)

            # Plotar as top-15 features mais importantes
            plt.figure(figsize=(12, 8))
            top_features = feature_importance_df.head(15)
            sns.barplot(x='importance', y='feature', data=top_features)
            plt.title(f'Top 15 Features Mais Importantes - {best_model_name}')
            plt.tight_layout()
            plt.show()

            print("   Top 10 features mais importantes:")
            for i, row in feature_importance_df.head(10).iterrows():
                print(f"   {i+1}. {row['feature']}: {row['importance']:.4f}")

        else:
            print(f"   O modelo {best_model_name} não fornece importância de features diretamente.")

            # Tentar permutation importance
            print("   Calculando importância por permutação...")
            perm_importance = permutation_importance(
                best_model, self.X_train_prepared, self.y_train_encoded,
                n_repeats=5, random_state=RANDOM_STATE, n_jobs=-1
            )

            # Obter nomes das features
            numeric_features = self.X_train.select_dtypes(include=np.number).columns.tolist()
            categorical_features = self.X_train.select_dtypes(exclude=np.number).columns.tolist()

            ohe = self.preprocessing_pipeline.named_transformers_['cat'].named_steps['onehot']
            categorical_features_encoded = ohe.get_feature_names_out(categorical_features)

            all_feature_names = numeric_features + list(categorical_features_encoded)

            # Criar DataFrame com importâncias
            sorted_idx = perm_importance.importances_mean.argsort()[::-1]
            top_idx = sorted_idx[:10]

            print("   Top 10 features por importância de permutação:")
            for i, idx in enumerate(top_idx):
                print(f"   {i+1}. {all_feature_names[idx]}: {perm_importance.importances_mean[idx]:.4f}")

    def _step5_4_analyze_types_of_errors(self):
        print("\n--- ETAPA 5.4: Analisando os tipos de erros que os modelos cometem ---")
        print("Investigando padrões de erro: Vou examinar a matriz de confusão do melhor modelo para entender que tipos de erros ele está cometendo.")

        # Treinar o melhor modelo
        best_model_name = self.model_comparison_results.index[0]
        best_model = self.models_to_test[best_model_name]
        best_model.fit(self.X_train_prepared, self.y_train_encoded)

        # Fazer previsões no conjunto de treino
        y_train_pred = best_model.predict(self.X_train_prepared)

        # Calcular matriz de confusão
        cm = confusion_matrix(self.y_train_encoded, y_train_pred)

        # Plotar matriz de confusão
        plt.figure(figsize=(10, 8))
        class_names = list(self.label_encoder_mapping.keys())
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=class_names, yticklabels=class_names)
        plt.title(f'Matriz de Confusão - {best_model_name} (Treino)')
        plt.ylabel('Verdadeiro')
        plt.xlabel('Previsto')
        plt.xticks(rotation=45)
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()

        print("   Análise de erros:")
        print("   - Os valores na diagonal representam classificações corretas.")
        print("   - Valores fora da diagonal representam erros de classificação.")
        print("   - Devo prestar atenção a padrões específicos de confusão entre classes.")

    def _step5_5_perform_quick_feature_selection_round(self):
        print("\n--- ETAPA 5.5: Realizando uma rodada rápida de seleção de features ---")
        print("Otimizando o conjunto de features: Com base na análise de importância, vou tentar remover features menos importantes e ver se o desempenho melhora.")

        # Esta é uma etapa opcional que pode ser realizada em iterações posteriores
        print("   Para este exemplo, vou pular a seleção de features detalhada.")
        print("   Manterei todas as features para a próxima etapa.")

    def _step5_6_perform_one_more_quick_iteration(self):
        print("\n--- ETAPA 5.6: Realizando mais uma iteração rápida ---")
        print("Iterando para melhorar: Este é o ciclo da ciência de dados - explorar, modelar, analisar, repetir.")
        print("   Com base nas análises anteriores, poderia:")
        print("   1. Experimentar diferentes técnicas de pré-processamento")
        print("   2. Balancear as classes com oversampling/undersampling")
        print("   3. Engenharia de features mais elaborada")
        print("   Para este exemplo, vou prosseguir com os modelos atuais.")

    def _step5_7_shortlist_top_models(self):
        print("\n--- ETAPA 5.7: Selecionando os modelos mais promissores ---")
        print("Selecionando os melhores candidatos: Com base nos resultados da validação cruzada, vou escolher os top 3 modelos para ajuste fino.")

        # Selecionar os top 3 modelos
        top_models = self.model_comparison_results.head(3)
        self.top_model_names = top_models.index.tolist()

        print(f"   Modelos selecionados para ajuste fino: {self.top_model_names}")

        for i, model_name in enumerate(self.top_model_names):
            score = top_models.loc[model_name, 'mean_f1']
            print(f"   {i+1}. {model_name}: F1-Score = {score:.4f}")

        self.best_model_name = self.top_model_names[0]
        print(f"   O modelo principal para focar será: {self.best_model_name}")

    # ==============================================================================
    # ETAPA 6: AJUSTAR FINAMENTE O SISTEMA
    # ==============================================================================
    def _step6_1_fine_tune_hyperparameters(self):
        print(f"\n--- ETAPA 6.1: Ajustando finamente os hiperparâmetros do {self.best_model_name} ---")
        print("Otimizando hiperparâmetros: Vou usar busca aleatória para encontrar a melhor combinação de hiperparâmetros para o modelo selecionado.")

        # Definir espaço de parâmetros para cada tipo de modelo
        param_grids = {
            'Random Forest': {
                'n_estimators': [100, 200, 300],
                'max_depth': [10, 20, 30, None],
                'min_samples_split': [2, 5, 10],
                'min_samples_leaf': [1, 2, 4],
                'bootstrap': [True, False]
            },
            'XGBoost': {
                'n_estimators': [100, 200, 300],
                'max_depth': [3, 5, 7],
                'learning_rate': [0.01, 0.1, 0.2],
                'subsample': [0.7, 0.8, 1.0],
                'colsample_bytree': [0.7, 0.8, 1.0]
            },
            'LightGBM': {
                'n_estimators': [100, 200, 300],
                'max_depth': [3, 5, 7, -1],  # -1 significa sem limite
                'learning_rate': [0.01, 0.1, 0.2],
                'num_leaves': [31, 50, 100],
                'subsample': [0.7, 0.8, 1.0]
            },
            'Logistic Regression': {
                'C': [0.1, 1, 10, 100],
                'solver': ['liblinear', 'saga'],
                'penalty': ['l1', 'l2']
            },
            'Gradient Boosting': {
                'n_estimators': [100, 200, 300],
                'max_depth': [3, 5, 7],
                'learning_rate': [0.01, 0.1, 0.2],
                'subsample': [0.7, 0.8, 1.0]
            }
        }

        # Obter o modelo base
        base_model = self.models_to_test[self.best_model_name]

        # Obter grade de parâmetros para este modelo
        if self.best_model_name in param_grids:
            param_grid = param_grids[self.best_model_name]
        else:
            # Grade de parâmetros genérica para outros modelos
            param_grid = {
                'C': [0.1, 1, 10, 100] if hasattr(base_model, 'C') else {},
                'max_depth': [3, 5, 7, 10] if hasattr(base_model, 'max_depth') else {},
                'n_estimators': [100, 200, 300] if hasattr(base_model, 'n_estimators') else {}
            }

        # Configurar busca aleatória
        kf = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)
        random_search = RandomizedSearchCV(
            base_model, param_distributions=param_grid, n_iter=20,
            scoring='f1_weighted', cv=kf, n_jobs=-1,
            random_state=RANDOM_STATE, verbose=1
        )

        print(f"   Iniciando busca de hiperparâmetros para {self.best_model_name}...")
        start_time = time.time()
        random_search.fit(self.X_train_prepared, self.y_train_encoded)
        search_time = time.time() - start_time

        print(f"   Busca concluída em {search_time:.2f} segundos.")
        print(f"   Melhores parâmetros: {random_search.best_params_}")
        print(f"   Melhor score na validação cruzada: {random_search.best_score_:.4f}")

        self.tuned_model = random_search.best_estimator_

    def _step6_2_try_ensemble_methods(self):
        print("\n--- ETAPA 6.2: Tentando métodos de Ensemble ---")
        print("Experimentando com ensembles: Vou combinar os melhores modelos em um ensemble para potencialmente melhorar o desempenho.")

        # Selecionar os top modelos para o ensemble
        ensemble_models = []
        for model_name in self.top_model_names[:3]:  # Usar top 3 modelos
            model = self.models_to_test[model_name]

            # Se for o modelo principal, usar a versão tunada
            if model_name == self.best_model_name:
                model = self.tuned_model

            ensemble_models.append((model_name, model))

        # Criar Voting Classifier
        voting_clf = VotingClassifier(
            estimators=ensemble_models,
            voting='soft',  # Usar probabilidades para votação
            n_jobs=-1
        )

        print("   Treinando ensemble com os modelos:")
        for name, _ in ensemble_models:
            print(f"   - {name}")

        # Avaliar ensemble com validação cruzada
        kf = StratifiedKFold(n_splits=3, shuffle=True, random_state=RANDOM_STATE)
        ensemble_scores = cross_val_score(
            voting_clf, self.X_train_prepared, self.y_train_encoded,
            cv=kf, scoring='f1_weighted', n_jobs=-1
        )

        ensemble_mean_score = np.mean(ensemble_scores)
        print(f"   Desempenho do ensemble: F1-Score = {ensemble_mean_score:.4f}")

        # Comparar com o melhor modelo individual
        best_individual_score = self.model_comparison_results.loc[self.best_model_name, 'mean_f1']
        print(f"   Desempenho do melhor modelo individual ({self.best_model_name}): {best_individual_score:.4f}")

        # Decidir se usar o ensemble ou o modelo individual
        if ensemble_mean_score > best_individual_score + 0.01:  # Melhoria significativa
            print("   O ensemble performou melhor. Usarei o ensemble como modelo final.")
            self.tuned_model = voting_clf
            self.best_model_name = "Ensemble"
        else:
            print("   O ensemble não ofereceu melhoria significativa. Manterei o modelo individual.")

        # Treinar o modelo final em todos os dados de treino
        print("   Treinando o modelo final em todos os dados de treino...")
        self.tuned_model.fit(self.X_train_prepared, self.y_train_encoded)

    def _step6_3_measure_performance_on_test_set(self):
        print("\n--- ETAPA 6.3: Medindo o desempenho no conjunto de teste (O Momento da Verdade) ---")
        print("Avaliação final: Agora, e somente agora, vou usar o modelo final ajustado para fazer previsões no conjunto de teste que separei no início.")

        # Fazer previsões no conjunto de teste
        self.test_set_predictions = self.tuned_model.predict(self.X_test_prepared)

        # Calcular métricas de desempenho
        self.final_f1_score = f1_score(self.y_test_encoded, self.test_set_predictions, average='weighted')
        final_accuracy = accuracy_score(self.y_test_encoded, self.test_set_predictions)
        final_precision = precision_score(self.y_test_encoded, self.test_set_predictions, average='weighted')
        final_recall = recall_score(self.y_test_encoded, self.test_set_predictions, average='weighted')

        print(f"   Desempenho no conjunto de teste:")
        print(f"   - F1-Score (Weighted): {self.final_f1_score:.4f}")
        print(f"   - Acurácia: {final_accuracy:.4f}")
        print(f"   - Precisão: {final_precision:.4f}")
        print(f"   - Recall: {final_recall:.4f}")

        print("\n   AVISO IMPORTANTE: A partir deste ponto, não devo mais fazer ajustes no modelo.")
        print("   Tentar melhorar o desempenho no conjunto de teste agora levaria ao overfitting.")

    # ==============================================================================
    # ETAPA 7: APRESENTAR A SOLUÇÃO
    # ==============================================================================
    def _step7_1_document_what_i_have_done(self):
        print("\n--- ETAPA 7.1: Documentando o que foi feito ---")
        print("Criando documentação completa: Resumo do projeto - Construí um pipeline completo que:")
        print("   1. Define claramente o problema de negócio e objetivos")
        print("   2. Obtém e prepara os dados de forma reprodutível")
        print("   3. Realiza análise exploratória detalhada")
        print("   4. Preprocessa os dados com técnicas apropriadas")
        print("   5. Compara múltiplos algoritmos de machine learning")
        print("   6. Ajusta os hiperparâmetros dos modelos mais promissores")
        print("   7. Avalia o modelo final em um conjunto de teste separado")
        print(f"   8. Atinge um F1-Score de {self.final_f1_score:.4f} no conjunto de teste")

    def _step7_2_create_a_nice_presentation(self):
        print("\n--- ETAPA 7.2: Criando uma apresentação visual ---")
        print("Preparando visualizações para apresentação: Vou criar gráficos e tabelas que destacam os resultados mais importantes.")

        # 1. Matriz de confusão
        print("\n   1. Matriz de Confusão Final:")
        cm = confusion_matrix(self.y_test_encoded, self.test_set_predictions)
        plt.figure(figsize=(10, 8))
        class_names = list(self.label_encoder_mapping.keys())
        sns.heatmap(cm, annot=True, fmt='d', cmap='Blues',
                   xticklabels=class_names, yticklabels=class_names)
        plt.title('Matriz de Confusão - Conjunto de Teste')
        plt.ylabel('Verdadeiro')
        plt.xlabel('Previsto')
        plt.xticks(rotation=45)
        plt.yticks(rotation=0)
        plt.tight_layout()
        plt.show()

        # 2. Relatório de classificação
        print("\n   2. Relatório de Classificação Detalhado:")
        print(classification_report(self.y_test_encoded, self.test_set_predictions,
                                  target_names=class_names))

        # 3. Comparação de modelos
        print("\n   3. Comparação do Desempenho dos Modelos:")
        plt.figure(figsize=(12, 6))
        models = self.model_comparison_results.index
        means = self.model_comparison_results['mean_f1']

        bars = plt.bar(range(len(models)), means)
        plt.xlabel('Modelos')
        plt.ylabel('F1-Score (Weighted)')
        plt.title('Comparação de Desempenho dos Modelos (Validação Cruzada)')
        plt.xticks(range(len(models)), models, rotation=45, ha='right')

        # Destacar o modelo escolhido
        chosen_idx = list(models).index(self.best_model_name)
        bars[chosen_idx].set_color('green')

        plt.tight_layout()
        plt.show()

    def _step7_3_explain_why_solution_achieves_objective(self):
        print("\n--- ETAPA 7.3: Explicando por que a solução atinge o objetivo de negócio ---")
        print("Conectando resultados técnicos com objetivos de negócio:")
        print(f"   O modelo final alcançou um F1-Score de {self.final_f1_score:.4f} no conjunto de teste.")

        if self.final_f1_score >= 0.85:
            print("   Este desempenho está acima do mínimo estabelecido de 85%, indicando que a solução é robusta e confiável o suficiente para ser usada como uma ferramenta de triagem automática.")
            print("   A implementação desta solução permitirá:")
            print("   - Redução do tempo de resposta a incidentes")
            print("   - Melhor priorização de alertas de segurança")
            print("   - Liberação de analistas para tarefas mais complexas")
            print("   - Maior consistência na classificação de ameaças")
        else:
            print("   Este desempenho está abaixo do mínimo estabelecido. A solução pode precisar de melhorias antes de ser implantada em produção.")

    def _step7_4_present_interesting_points(self):
        print("\n--- ETAPA 7.4: Apresentando pontos interessantes observados ---")
        print("Destacando insights valiosos descobertos durante a análise:")
        print("   - O que funcionou bem:")
        print("     * A validação cruzada foi essencial para comparar modelos objetivamente")
        print("     * O pré-processamento adequado melhorou significativamente o desempenho")
        print("     * Ensemble methods proporcionaram ganhos modestos mas consistentes")
        print("   - Desafios encontrados:")
        print("     * O desbalanceamento de classes exigiu métricas apropriadas (F1-Score)")
        print("     * Features com alta cardinalidade precisaram de tratamento especial")
        print("     * Valores ausentes em algumas colunas requereram estratégias de imputação")
        print("   - Limitações do modelo:")
        print("     * O modelo está limitado aos tipos de ataque presentes nos dados históricos")
        print("     * Novas ameaças (zero-day) podem não ser classificadas corretamente")
        print("     * O desempenho pode variar para diferentes contextos organizacionais")

    def _step7_5_ensure_key_findings_are_communicated(self):
        print("\n--- ETAPA 7.5: Comunicando os achados principais de forma simples ---")
        print("Síntese executiva dos resultados:")
        print("   **CONCLUSÃO PRINCIPAL**: Desenvolvemos um sistema de classificação automática de ciberameaças que alcança alta precisão na categorização de incidentes de segurança.")
        print("   **FATORES-CHAVE DE SUCESSO**: ")
        print("   - Uso de algoritmos ensemble (especialmente Gradient Boosting)")
        print("   - Pré-processamento cuidadoso dos dados")
        print("   - Seleção adequada de features mais relevantes")
        print("   **PRÓXIMOS PASSOS RECOMENDADOS**: ")
        print("   - Implementar monitoramento contínuo do desempenho")
        print("   - Estabelecer processo de retreinamento regular com novos dados")
        print("   - Desenvolver interface para analistas revisarem e corrigirem classificações")

    # ==============================================================================
    # ETAPA 8: LANÇAR, MONITORAR E MANTER
    # ==============================================================================
    def _step8_1_get_solution_ready_for_production(self):
        print("\n--- ETAPA 8.1: Preparando a solução para produção ---")
        print("Empacotando a solução para implantação: Vou salvar o pipeline completo (pré-processamento + modelo) para facilitar a implantação.")

        # Criar pipeline completo para produção
        production_pipeline = Pipeline([
            ('preprocessing', self.preprocessing_pipeline),
            ('model', self.tuned_model)
        ])

        # Treinar o pipeline completo nos dados de treino
        production_pipeline.fit(self.X_train, self.y_train)

        # Salvar o pipeline para uso futuro
        joblib.dump(production_pipeline, 'cybersecurity_threat_classifier.pkl')
        print("   Pipeline salvo como 'cybersecurity_threat_classifier.pkl'")

        # Também salvar o label encoder para decodificar previsões
        joblib.dump(self.label_encoder_mapping, 'label_encoder_mapping.pkl')
        print("   Mapeamento de labels salvo como 'label_encoder_mapping.pkl'")

        print("   Para usar o modelo em produção:")
        print("   1. Carregar o pipeline salvo")
        print("   2. Pré-processar novos dados da mesma forma")
        print("   3. Fazer previsões com model.predict()")
        print("   4. Decodificar labels se necessário")

    def _step8_2_write_monitoring_code(self):
        print("\n--- ETAPA 8.2: Escrevendo código de monitoramento ---")
        print("Preparando monitoramento contínuo: Um modelo em produção precisa ser monitorado para detectar degradação de desempenho.")

        monitoring_code = """
        # Exemplo de código para monitoramento do modelo
        def monitor_model_performance(new_data, new_labels, pipeline_path='cybersecurity_threat_classifier.pkl'):
            # Carregar pipeline
            pipeline = joblib.load(pipeline_path)

            # Fazer previsões
            predictions = pipeline.predict(new_data)

            # Calcular métricas de desempenho
            f1 = f1_score(new_labels, predictions, average='weighted')
            accuracy = accuracy_score(new_labels, predictions)

            # Verificar se o desempenho caiu abaixo de um threshold
            if f1 < 0.8:  # Threshold arbitrário
                print(f"ALERTA: Desempenho do modelo caiu para F1-Score={f1:.4f}")
                # Disparar retreinamento ou investigação

            return f1, accuracy

        # Também monitorar a distribuição dos dados de entrada
        def monitor_data_drift(new_data, training_data):
            # Comparar distribuição dos dados novos com dados de treino
            # Usar teste Kolmogorov-Smirnov para features numéricas
            # Usar divergência de Jensen-Shannon para features categóricas
            pass
        """

        print("   Código de monitoramento conceitual:")
        print(monitoring_code)
        print("   Na implementação real, isso seria integrado ao pipeline de CI/CD e sistema de alertas.")

    def _step8_3_retrain_models_on_a_regular_basis(self):
        print("\n--- ETAPA 8.3: Retreinando os modelos regularmente ---")
        print("Planejando manutenção contínua: O mundo da cibersegurança evolui rapidamente, então o modelo precisa ser retreinado periodicamente.")

        retraining_plan = """
        Plano de Retreinamento:

        1. Frequência: A cada 3 meses, ou quando:
           - O desempenho monitorado cair abaixo do threshold
           - Novos tipos de ataques forem identificados
           - Mudanças significativas na infraestrutura ocorrerem

        2. Processo:
           a. Coletar novos dados rotulados
           b. Validar qualidade dos novos dados
           c. Adicionar aos dados históricos
           d. Reexecutar o pipeline completo de treinamento
           e. Validar o novo modelo contra o atual
           f. Implantar se houver melhoria significativa

        3. Versionamento:
           - Manter versões dos modelos implantados
           - Manter registro do desempenho de cada versão
           - Ter capacidade de rollback se necessário
        """

        print("   Plano de retreinamento:")
        print(retraining_plan)
        print("   Este processo deve ser automatizado tanto quanto possível para garantir consistência.")

# ==============================================================================
# PONTO DE ENTRADA PRINCIPAL
# ==============================================================================
# Agora que defini todo o meu plano de ação, vou criar uma instância da minha
# classe e executar o pipeline completo.
# ==============================================================================
if __name__ == '__main__':
    print("Iniciando análise de ameaças de cibersegurança...")
    print("=" * 80)

    analysis_pipeline = CybersecurityThreatAnalysisPipeline()
    analysis_pipeline.run_full_pipeline()
```
